distance table data cleaning

1)

Data Cleaning to delete the rows with same start and end timestamps if any exists

```python
same_time = distance[
    distance["start_timestamp"] == distance["end_timestamp"]
]

print(f"Number of records with same start and end timestamps: {len(same_time)}")

```

```python
distance = distance[
    distance["start_timestamp"] != distance["end_timestamp"]
].copy()
```

```python
distance.shape
```

```python
import pandas as pd
#converting start_timestamp and end_timestamp to datetime
distance["start_timestamp"] = pd.to_datetime(distance["start_timestamp"])
distance["end_timestamp"] = pd.to_datetime(distance["end_timestamp"])
distance[["start_timestamp", "end_timestamp"]].head()
#distance.dtypes
```

```python
distance.shape
```

```python
#calculating duration in minutes (start-end timestamps)

distance['duration'] = (distance['end_timestamp'] - distance['start_timestamp']).dt.total_seconds() / 60
distance[['start_timestamp', 'end_timestamp', 'duration']].head()
```

3)

Identify cases with start_time stamp before september 2025 (study timeframe)


```python
pre2025=distance[distance["start_timestamp"] < "2025-09-01"][["start_timestamp", "end_timestamp", "duration", "distance", "app_user_id", "app_source"]]


```

```python
distance[distance["start_timestamp"] < "2025-01-01"][["start_timestamp", "end_timestamp", "duration", "distance", "app_user_id", "app_source"]]
```

```python
pre2025.groupby("app_user_id").size().sort_values(ascending=False)
```

```python
#remove cases with start_time stamp before september 2025
distance_cleaned = distance[distance["start_timestamp"] >= "2025-09-01"]
distance_cleaned.shape
```

--------------------------------------------------------------

4)

Removing duplicate rows having the same [app_user_id, start_timestamp, end_timestamp, distance, app_source]


```python
duplicate_count = distance_cleaned.duplicated(
    subset=[
        "app_user_id",
        "start_timestamp",
        "end_timestamp",
        "distance",
        "app_source"
    ]
).sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
duplicates = distance_cleaned[
    distance_cleaned.duplicated(
        subset=[
            "app_user_id",
            "start_timestamp",
            "end_timestamp",
            "distance",
            "app_source"
        ],
        keep=False
    )
]

duplicates.sort_values(
    ["app_user_id", "start_timestamp"]
)
```

```python
duplicates.to_csv("cleaned_duplicate_rows_distance.csv", index=False)
```

```python
distance_cleaned = distance_cleaned.drop_duplicates(
    subset=[
        "app_user_id",
        "start_timestamp",
        "end_timestamp",
        "distance",
        "app_source"
    ],
    keep="first"
)

print("Shape after removing duplicates:", distance_cleaned.shape)
```

```python
#checking if we have any duplicates now:
duplicate_rows = distance_cleaned[
    distance_cleaned.duplicated(
        subset=[
            "app_user_id",
            "start_timestamp",
            "end_timestamp",
            "distance",
            "app_source"
        ],
        keep=False
    )
]

print("Number of duplicate rows:", len(duplicate_rows))
duplicate_rows
```
